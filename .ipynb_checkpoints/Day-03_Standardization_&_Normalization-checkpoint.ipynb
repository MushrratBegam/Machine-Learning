```python
# Standardization & Normalization

# Standardization:- center the data around mean(μ = 0) with a standerd division(σ = 1).(Range:-No fixed limits,Center :-centered at mean = 0,Outlieres:- less affected)
# Standardization (it is good to go with standardization if you have outlier in your dataset)
# z = x(Data)-μ(mean)/σ(std) 
```

```python
data = {
    'Empid':[x for x in range(101,111)],
    'Empsalary':[23445,78786,78987,932455,67554,78987,23244,67865,34554,67876]
}
```

```python
import pandas as pd
```

```python
# Scaling
# Salary : 10k, 1l, 10l /10l  => 0.01, 0.1, 1
# Age = 54, 67, 45 / 100      => 0.54, 0.67, 0.45
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df['scalesalary'] = (df['Empsalary']-df['Empsalary'].mean())/df['Empsalary'].std()
```

```python
df
```

```python
df.scalesalary.max()
```

```python
df = df[df.scalesalary<df.scalesalary.max()]
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
newdf = StandardScaler().fit_transform(df)
```

```python
df
```

```python
df.describe()
```

```python
# Normalization(Min-Max Scalling):- Rescale data into a fixed range,usually between 0 and 1 
#(Range:- fixed[0,1], Center:-Scalling based on min/max, Outliers:-Highly affected)
# z = (x-min) / (max-min)
```

```python
df = df[df['Empsalary']<500000]
```

```python
df
```

```python
df['scalesalary'] = (df['Empsalary'] - df['Empsalary'].min()) / (df['Empsalary'].max() - df['Empsalary'].min())
```

```python
df
```

```python
df = df.drop(columns=['scalesalary'])
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
df = MinMaxScaler().fit_transform(df)
```
